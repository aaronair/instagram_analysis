#2. Аналитика

Считаю среднее количество лайков, комментариев, хэштегов, ER, среднюю длину постов

```python
import numpy as np

#среднее кол-во лайков
def avg_likes(posts):

    likes = [
        p.get("likes", 0)
        for p in posts
    ]

    return np.mean(likes)
```

```python
#среднее кол-во комментариев

def avg_comments(posts):

    comments = [
        p.get("comments", 0)
        for p in posts
    ]

    return np.mean(comments)
```

```python
def calculate_er(data):

    followers = data.get(
        "followers",
        0
    )

    if followers == 0:
        return 0


    avg_like = avg_likes(
        data["posts"]
    )

    avg_comment = avg_comments(
        data["posts"]
    )


    return (
        (avg_like + avg_comment)
        /
        followers
        *
        100
    )
```

```python

def avg_caption_length(posts):

    lengths = []

    for p in posts:

        caption = p.get("caption") or ""

        lengths.append(
            len(caption)
        )

    return np.mean(lengths) if lengths else 0



def avg_hashtags(posts):

    counts = []

    for p in posts:

        hashtags = p.get("hashtags") or []

        counts.append(
            len(hashtags)
        )

    return np.mean(counts) if counts else 0
```

1. Доля рекламных постов

```python
AD_WORDS = [
    "реклама",
    "#реклама",
    "ad",
    "#ad",
    "партнерство",
    "партнёрство",
    "сотрудничество",
    "промокод",
    "по промокоду",
    "подарок",
    "подарили",
    "от бренда",
    "collab"
]
```

```python
def ad_ratio(posts):

    if not posts:
        return 0

    ad_posts = 0

    for p in posts:

        caption = (
            p.get("caption") or ""
        ).lower()


        if any(
            word in caption
            for word in AD_WORDS
        ):
            ad_posts += 1


    return ad_posts / len(posts)
```

2. Количество упоминаний брендов


```python
def avg_mentions(posts):

    values = []

    for p in posts:

        mentions = p.get(
            "mentions",
            []
        )

        values.append(
            len(mentions)
        )

    return np.mean(values) if values else 0
```

3. Частота публикаций


```python
from datetime import datetime


def posting_frequency(posts):

    if len(posts) < 2:
        return 0


    dates = []

    for p in posts:

        date = p.get(
            "timestamp"
        )

        if date:
            dates.append(
                datetime.fromisoformat(
                    date.replace("Z", "+00:00")
                )
            )


    if len(dates) < 2:
        return 0


    days = (
        max(dates) - min(dates)
    ).days


    if days == 0:
        return len(posts)


    return len(posts) / days * 30

```

4. Медианный ER
Средний ER может ломаться из-за одного вирусного ролика.



```python
def median_er(data):

    followers = data.get(
        "followers",
        0
    )

    if followers == 0:
        return 0


    ers = []


    for p in data["posts"]:

        likes = p.get(
            "likes",
            0
        )

        comments = p.get(
            "comments",
            0
        )


        er = (
            (likes + comments)
            /
            followers
            *
            100
        )

        ers.append(er)


    return np.median(ers) if ers else 0
```

5. Распределение форматов

Reels;
фото.


```python
def content_types(posts):

    total = len(posts)

    if total == 0:
        return {
            "video_ratio": 0,
            "photo_ratio": 0
        }


    videos = sum(
        1
        for p in posts
        if p.get("video_url")
    )


    return {

        "video_ratio":
            videos / total,

        "photo_ratio":
            (total - videos) / total
    }
```
